```python
#Konfidensintervall - Beräkna ett konfidensintervall för medelvärdet av systolic_bp (t.ex. med normalapproximation eller bootstrap, som vi gått igenom på kursen).
x = df["systolic_bp"].dropna().to_numpy()
n = len(x)
mean = x.mean()
sd = x.std(ddof=1)
se = sd / np.sqrt(n)

lower = mean - 1.96 * se
upper = mean + 1.96 * se

print("Punktestimat: ", mean)
print("Standardavvikelse på stickprov: ", sd)
print("Standardfel: ", se)
print("95% konfidensintervall: ", (float(lower), float(upper)))
```
